```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import LeaveOneOut
import numpy as np
```

```python
path = "/content/drive/MyDrive/Project_Machine_Learning/Praktikum_Mandiri_10/Data/"

df = pd.read_excel(path + "Raining.xlsx")
```

```python
df
```

```python
X = df[['Temperatur Udara (°C)', 'Kecepatan Angin (km/jam)']]
y = df['Klasifikasi atau Persepsi Marry']
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
loo = LeaveOneOut()
k_values = range(1, len(df))
accuracies = []
```

```python
for k in k_values:
    correct = 0
    model = KNeighborsClassifier(n_neighbors=k)

    for train_index, test_index in loo.split(X_scaled):
        X_train, X_test = X_scaled[train_index], X_scaled[test_index]
        y_train, y_test = y_encoded[train_index], y_encoded[test_index]

        model.fit(X_train, y_train)
        pred = model.predict(X_test)

        if pred == y_test:
            correct += 1

    accuracy = correct / len(df)
    accuracies.append(accuracy)
```

```python
for k, acc in zip(k_values, accuracies):
    print(f"k = {k}, Akurasi LOOCV = {acc:.3f}")
```

```python
best_k = k_values[np.argmax(accuracies)]
print("\nNilai k terbaik:", best_k)
```

```python
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(X_scaled, y_encoded)
```

```python
new_data = scaler.transform([[16, 3]])
prediction = knn.predict(new_data)
```

```python
print("Prediksi Persepsi Marry:", le.inverse_transform(prediction)[0])
```
